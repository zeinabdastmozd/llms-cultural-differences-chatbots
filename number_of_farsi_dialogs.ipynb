```python
import os
import json
import pandas as pd
from langdetect import detect

# ---------- Helper function to detect language ----------
def detect_language(text):
    """Detect language of the given text using langdetect."""
    try:
        lang = detect(text)
        if lang == 'fa':  # Farsi (Persian)
            return 'farsi'
        elif lang == 'en':  # English
            return 'english'
        else:
            return 'other'
    except:
        return 'unknown'

# ---------- Main function ----------
def summarize_languages(data_path):
    rows = []

    for root, dirs, files in os.walk(data_path):
        for file in files:
            if file.endswith(".json"):
                filepath = os.path.join(root, file)

                # Load JSON
                with open(filepath, "r", encoding="utf-8") as f:
                    content = json.load(f)

                # Combine all dialog text
                dialog_text = ""
                if "dialog" in content:
                    for turn in content["dialog"]:
                        dialog_text += " " + " ".join([str(v) for v in turn.values()])

                # Detect language
                lang = detect_language(dialog_text)
                english_count = 1 if lang == 'english' else 0
                farsi_count = 1 if lang == 'farsi' else 0

                # Extract info from file name: Cohabitation_temp_0.1_1_Daughter_Father_Agreement.json
                filename = file.replace(".json", "")
                parts = filename.split("_")

                topic = parts[0]                 # Cohabitation
                temperature = parts[2]           # 0.1
                # generation_number = int(parts[3])  # Not used
                # speakers and stance = ignored

                # Determine culture from folder structure
                culture = "UK" if "UK" in root else "Iran"

                # Get model name (the folder just before culture folder)
                model = os.path.basename(os.path.dirname(root))

                rows.append({
                    "model": model,
                    "culture": culture,
                    "topic": topic,
                    "temperature": temperature,
                    "english_count": english_count,
                    "farsi_count": farsi_count
                })

    # Create DataFrame
    df = pd.DataFrame(rows)
    return df

# ---------- Run ----------
data_path = "/home/shekhar/Desktop/code/persona/output2/Dissertation_All_Models"
df_summary = summarize_languages(data_path)

# Preview
print(df_summary.head())

```

```python
import os
import json
import pandas as pd
from langdetect import detect

# ---------- Helper function to detect language ----------
def detect_language(text):
    """Detect language of the given text using langdetect."""
    try:
        lang = detect(text)
        if lang == 'fa':
            return 'farsi'
        elif lang == 'en':
            return 'english'
        else:
            return 'other'
    except:
        return 'unknown'

# ---------- Modified Main Function ----------
def summarize_languages(data_path):
    rows = []

    for root, dirs, files in os.walk(data_path):
        for file in files:
            if file.endswith(".json"):
                filepath = os.path.join(root, file)

                # Load JSON
                with open(filepath, "r", encoding="utf-8") as f:
                    content = json.load(f)

                # Combine all dialog text
                dialog_text = ""
                if "dialog" in content:
                    for turn in content["dialog"]:
                        dialog_text += " " + " ".join([str(v) for v in turn.values()])

                # Detect language
                lang = detect_language(dialog_text)
                english_count = 1 if lang == 'english' else 0

                # Determine culture from folder structure
                culture = "UK" if "UK" in root else "Iran"
                model = os.path.basename(os.path.dirname(root))

                rows.append({
                    "model": model,
                    "culture": culture,
                    "english": english_count,
                    "total": 1
                })

    # Create DataFrame
    df = pd.DataFrame(rows)

    # Group by model and culture
    summary = df.groupby(["model", "culture"]).agg({
        "english": "sum",
        "total": "count"  # same as number of dialogs
    }).reset_index()

    summary.rename(columns={
        "english": "english_count",
        "total": "total_count"
    }, inplace=True)

    # Optionally add farsi count
    summary["farsi_count"] = summary["total_count"] - summary["english_count"]

    return summary
```

```python
df_summary = summarize_languages(data_path)
df_summary.to_csv("language_summary.csv", index=False)
```
